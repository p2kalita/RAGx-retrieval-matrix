## **Section 1 — Sample Size Justification**

## **1.1 Introduction**

An adequate sample size is essential to ensure that the statistical analyses conducted in this study have sufficient power to detect meaningful differences in retrieval and answer-generation performance. A sample that is too small may fail to identify genuine performance differences, whereas an unnecessarily large sample may increase annotation and computational effort without providing proportionate analytical benefit.

In this study, the unit of analysis is the evaluation question. Each question is processed using the same retrieval models and, where applicable, the same answer-generation framework. This repeated-measures design allows model performance to be compared on identical legal queries, thereby reducing variability arising from differences in question difficulty.

The required sample size is assessed separately for each research question because the statistical test, study design, outcome measure, and effect-size assumption differ across the evaluation tasks. The final evaluation sample size is determined by the largest minimum sample-size requirement obtained across RQ1–RQ4, ensuring that all planned statistical analyses are adequately powered.

The sample-size analysis is based on the following statistical criteria:

- **Significance level:** $\alpha = 0.05$
- **Statistical power:** $1-\beta = 0.80$
- **Two-sided hypothesis testing**
- **Medium effect size assumptions** where pilot estimates are unavailable.

The planned evaluation dataset contains 50 manually curated legal questions, including both answerable and unanswerable queries. The adequacy of this sample will be assessed against the minimum sample-size requirement calculated for each research question.

## **1.2 Research Questions and Statistical Analysis Plan**

The proposed **NYAYaRAG** framework is evaluated through four research questions addressing three complementary components of retrieval-augmented legal question answering: **retrieval performance**, **evidence-grounded answer generation**, and **safe abstention**. Each research question is associated with an appropriate statistical procedure and corresponding evaluation metrics selected according to the study design, outcome variable, and measurement scale.

**RQ1** evaluates retrieval performance at the retrieval-paradigm level by comparing semantic retrieval models with traditional lexical retrieval models. Since each benchmark legal research question is evaluated under both retrieval paradigms, the observations are paired. Accordingly, a **one-sided Wilcoxon signed-rank test** is employed to determine whether semantic retrieval significantly outperforms lexical retrieval. Retrieval performance is evaluated using **Recall@k, Mean Reciprocal Rank (MRR), nDCG@k, Hit Rate@k, paired lexical-semantic score differences, p-values, and rank-biserial correlation** as the effect-size measure.

**RQ2** compares the retrieval performance of the six individual retrieval methodologies (TF-IDF, BM25, E5, BGE, InLegalBERT, and Hybrid Retrieval). As each benchmark question is evaluated using all six retrieval models, the study follows a repeated-measures design. Consequently, the **Friedman test** is used to detect overall differences among the retrieval models. When the Friedman test is statistically significant, **Wilcoxon signed-rank post-hoc comparisons** with **Holm adjustment** are performed to identify the specific retrieval models that differ significantly. Retrieval performance is assessed using **Recall@k, MRR, nDCG@k, Hit Rate@k, Friedman mean ranks, Holm-adjusted p-values, and appropriate effect sizes**.

**RQ3** evaluates the quality of evidence-grounded answers generated by NYAYaRAG. Each benchmark legal research question produces a single composite answer-quality score derived from predefined evaluation criteria. The observed answer-quality scores are first summarised descriptively and are then compared against a predefined minimum acceptable benchmark using a **one-sample Wilcoxon signed-rank test**. The principal evaluation metrics include **evidence-groundedness, legal correctness, citation accuracy, answer relevance, completeness, and the composite answer-quality score**.

**RQ4** evaluates the ability of NYAYaRAG to correctly abstain from answering legal research questions when the retrieved judgments do not provide sufficient supporting evidence. Abstention performance is summarised using **correct-abstention rate, abstention precision, abstention recall, F1-score, false-answer rate, false-abstention rate, and hallucination rate**. The observed correct-abstention rate is subsequently evaluated using an **exact one-sided binomial test** to determine whether it significantly exceeds a predefined minimum acceptable abstention benchmark.

## **1.3 Sample Size Determination**


An appropriate sample size is essential to ensure that the statistical analyses performed in this study have sufficient power to detect meaningful differences in retrieval performance and answer-generation quality. An inadequate sample size increases the probability of Type II errors, whereas an excessively large sample may result in unnecessary computational effort and manual annotation without providing proportionate analytical benefit.

The required sample size depends on the statistical test employed, the expected effect size, the desired significance level, and the statistical power. Since the four research questions investigate different aspects of retrieval performance, answer generation, and abstention, separate prospective sample-size calculations are performed for each research question.

The sample-size calculations presented in this section are based on the following statistical assumptions:

- **Significance level (α):** 0.05

## **1.4 Sample Size Determination for RQ1**

### **1.4.1 Statistical Basis**

Research Question 1 evaluates whether **semantic retrieval models significantly outperform traditional lexical retrieval models** for legal information retrieval over Indian Supreme Court judgments.

The retrieval paradigms considered in this comparison comprise:

- **Lexical retrieval models:** TF-IDF and BM25
- **Semantic retrieval models:** E5, BGE, and InLegalBERT

Each benchmark legal research question is evaluated using both retrieval paradigms. Consequently, every question produces a paired lexical retrieval score and semantic retrieval score. Because the observations originate from the same benchmark questions, they are related rather than independent, resulting in a **paired repeated-measures experimental design**.

The primary statistical test selected for RQ1 is the **Wilcoxon signed-rank test**. This non-parametric test is appropriate for comparing two related samples when the distribution of paired differences cannot be assumed to be normal. Retrieval-performance measures such as Recall@k, Mean Reciprocal Rank (MRR), and nDCG@k are frequently bounded, skewed, and may contain tied observations. The Wilcoxon signed-rank test therefore provides a robust statistical procedure for evaluating whether semantic retrieval consistently produces higher retrieval-performance scores than lexical retrieval.

Because the research hypothesis is directional and specifically proposes that semantic retrieval outperforms lexical retrieval, a **one-sided Wilcoxon signed-rank test** is adopted.

The hypotheses for RQ1 are defined as follows.

**Research Question 1**

- **Null Hypothesis (H₀₁):** Semantic retrieval models do not provide significantly higher retrieval performance than traditional lexical retrieval models for legal information retrieval over Indian Supreme Court judgments.

- **Alternative Hypothesis (H₁₁):** Semantic retrieval models provide significantly higher retrieval performance than traditional lexical retrieval models for legal information retrieval over Indian Supreme Court judgments.

Unlike the Friedman test, the Wilcoxon signed-rank test is based on paired observations. Consequently, the required sample size is determined using a prospective power analysis for a paired non-parametric comparison.

The prospective sample-size analysis assumes:

- **Significance level (α):** 0.05
- **Statistical power (1 − β):** 0.80
- **One-sided hypothesis test**
- **Conservative prospective effect-size assumption** in the absence of pilot-derived estimates.

The required sample size is identified as the smallest number of benchmark legal research questions that achieves the target statistical power under these assumptions. This minimum requirement will subsequently be compared with the corresponding requirements for RQ2, RQ3, and RQ4, and the largest minimum sample size will be adopted as the final benchmark evaluation sample.

### **Table 1. Statistical Design and Power Analysis Parameters for Retrieval Evaluation (RQ1)**

| **Parameter** | **Specification** |
|:---|:---|
| **Applicable Research Question** | RQ1 |
| **Retrieval Paradigms Compared** | Lexical (TF-IDF, BM25) vs Semantic (E5, BGE, InLegalBERT) |
| **Experimental Unit** | Individual benchmark legal research question |
| **Study Design** | Paired repeated-measures design |
| **Primary Statistical Test** | One-sided Wilcoxon signed-rank test |
| **Significance Level ($\alpha$)** | 0.05 |
| **Target Statistical Power ($1-\beta$)** | 0.80 |
| **Effect Size Measure** | Rank-biserial correlation (or equivalent Wilcoxon effect-size measure) |
| **Effect Size Assumption** | Conservative prospective effect-size assumption |
| **Planned Evaluation Sample** | To be determined through prospective sample-size analysis |

Because this analysis is conducted before the retrieval evaluation, the assumed effect size is used solely for prospective sample-size planning. Following completion of the retrieval evaluation, the Wilcoxon signed-rank test statistic, the corresponding *p*-value, and the rank-biserial correlation effect size will be reported based on the observed retrieval performance.

The minimum sample-size estimate obtained from this analysis applies only to **RQ1**, which evaluates the retrieval performance of semantic and lexical retrieval paradigms. The final benchmark evaluation sample will be selected after comparing the minimum sample-size requirements across **RQ1–RQ4**, with the largest requirement adopted for the study.

```python
# ============================================================
# RQ1: PROSPECTIVE SAMPLE-SIZE ANALYSIS
# FOR A PAIRED LEXICAL–SEMANTIC RETRIEVAL COMPARISON
# ============================================================

from statsmodels.stats.power import TTestPower
from IPython.display import display


# ------------------------------------------------------------
# Step 1: Define the power-analysis parameters
# ------------------------------------------------------------

ALPHA = 0.05
TARGET_POWER = 0.80

# RQ1 contains a directional hypothesis:
# semantic retrieval performance > lexical retrieval performance
TEST_ALTERNATIVE = "larger"

# Conservative medium standardized paired effect-size assumption
# d_z represents the mean paired difference divided by the
# standard deviation of the paired differences.
ASSUMED_EFFECT_SIZE_DZ = 0.50

# Candidate numbers of benchmark legal research questions
SAMPLE_SIZES = range(5, 101)

# Planned evaluation sample
PLANNED_SAMPLE_SIZE = 50


# ------------------------------------------------------------
# Step 2: Initialise the paired-design power model
# ------------------------------------------------------------

power_model = TTestPower()


# ------------------------------------------------------------
# Step 3: Calculate power across candidate sample sizes
# ------------------------------------------------------------

power_results = []

for n_questions in SAMPLE_SIZES:

    estimated_power = power_model.power(
        effect_size=ASSUMED_EFFECT_SIZE_DZ,
        nobs=n_questions,
        alpha=ALPHA,
        alternative=TEST_ALTERNATIVE
    )

    power_results.append({
        "Evaluation Questions": n_questions,
        "Assumed Effect Size (d_z)": ASSUMED_EFFECT_SIZE_DZ,
        "Estimated Power": estimated_power,
        "Target Power Achieved": estimated_power >= TARGET_POWER
    })


rq1_power_results = pd.DataFrame(power_results)


# ------------------------------------------------------------
# Step 4: Identify the minimum required sample size
# ------------------------------------------------------------

adequate_sample_sizes = rq1_power_results[
    rq1_power_results["Estimated Power"] >= TARGET_POWER
]

if adequate_sample_sizes.empty:
    minimum_required_n = None
else:
    minimum_required_n = int(
        adequate_sample_sizes.iloc[0]["Evaluation Questions"]
    )


# ------------------------------------------------------------
# Step 5: Confirm the result using solve_power
# ------------------------------------------------------------

analytical_required_n = power_model.solve_power(
    effect_size=ASSUMED_EFFECT_SIZE_DZ,
    nobs=None,
    alpha=ALPHA,
    power=TARGET_POWER,
    alternative=TEST_ALTERNATIVE
)

rounded_required_n = int(-(-analytical_required_n // 1))


# ------------------------------------------------------------
# Step 6: Extract power for the planned sample of 50 questions
# ------------------------------------------------------------

planned_sample_power = float(
    rq1_power_results.loc[
        rq1_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE,
        "Estimated Power"
    ].iloc[0]
)


# ------------------------------------------------------------
# Step 7: Validation checks
# ------------------------------------------------------------

assert 0 < ALPHA < 1, (
    "Alpha must be between 0 and 1."
)

assert 0 < TARGET_POWER < 1, (
    "Target power must be between 0 and 1."
)

assert ASSUMED_EFFECT_SIZE_DZ > 0, (
    "The assumed effect size must be greater than zero."
)

assert TEST_ALTERNATIVE in ["two-sided", "larger", "smaller"], (
    "The specified test direction is invalid."
)

assert PLANNED_SAMPLE_SIZE in rq1_power_results[
    "Evaluation Questions"
].values, (
    "The planned sample size is not included in the candidate range."
)

assert rq1_power_results["Estimated Power"].between(0, 1).all(), (
    "All estimated power values must be between 0 and 1."
)

assert rq1_power_results["Estimated Power"].is_monotonic_increasing, (
    "Estimated power should increase as sample size increases."
)

assert rq1_power_results.isna().sum().sum() == 0, (
    "The results table contains missing values."
)

assert minimum_required_n == rounded_required_n, (
    "The iterative and analytical sample-size estimates do not agree."
)


# ------------------------------------------------------------
# Step 8: Display the most relevant rows
# ------------------------------------------------------------

if minimum_required_n is not None:

    lower_limit = max(
        min(SAMPLE_SIZES),
        minimum_required_n - 3
    )

    upper_limit = minimum_required_n + 3

    nearby_rows = rq1_power_results[
        rq1_power_results["Evaluation Questions"].between(
            lower_limit,
            upper_limit
        )
    ]

    planned_row = rq1_power_results[
        rq1_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE
    ]

    display_table = pd.concat(
        [nearby_rows, planned_row],
        ignore_index=True
    ).drop_duplicates(
        subset=["Evaluation Questions"]
    )

else:
    display_table = rq1_power_results.tail(10)


display(
    display_table.style
    .hide(axis="index")
    .format({
        "Assumed Effect Size (d_z)": "{:.2f}",
        "Estimated Power": "{:.3f}"
    })
    .set_caption(
        "RQ1 Prospective Power Analysis for the "
        "Paired Lexical–Semantic Retrieval Comparison"
    )
    .set_properties(**{
        "text-align": "center",
        "vertical-align": "middle"
    })
)


# ------------------------------------------------------------
# Step 9: Print the final sample-size conclusion
# ------------------------------------------------------------

print("=" * 72)
print("RQ1 SAMPLE-SIZE DETERMINATION")
print("=" * 72)

print(
    f"Study design                       : "
    f"Paired lexical-semantic comparison"
)

print(
    f"Planned final statistical test     : "
    f"One-sided Wilcoxon signed-rank test"
)

print(
    f"Prospective planning approximation : "
    f"Paired-samples normal approximation"
)

print(
    f"Significance level                 : "
    f"{ALPHA:.2f}"
)

print(
    f"Target statistical power           : "
    f"{TARGET_POWER:.2f}"
)

print(
    f"Assumed standardized effect size   : "
    f"d_z = {ASSUMED_EFFECT_SIZE_DZ:.2f}"
)

print(
    f"Direction of hypothesis            : "
    f"Semantic retrieval > lexical retrieval"
)


if minimum_required_n is not None:

    print(
        f"\nMinimum sample size achieving power ≥ "
        f"{TARGET_POWER:.2f}: "
        f"{minimum_required_n} questions"
    )

    print(
        f"Unrounded analytical estimate      : "
        f"{analytical_required_n:.3f}"
    )

    print(
        f"Estimated power with "
        f"{PLANNED_SAMPLE_SIZE} questions      : "
        f"{planned_sample_power:.3f}"
    )

    if PLANNED_SAMPLE_SIZE >= minimum_required_n:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions exceeds "
            f"the estimated minimum sample-size requirement for RQ1."
        )

    else:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions does not meet "
            f"the estimated minimum sample-size requirement for RQ1."
        )

else:

    print(
        "\nThe target statistical power was not achieved "
        "within the candidate sample-size range."
    )


print(
    "\nNote: The prospective calculation uses a paired-samples "
    "normal approximation because a universally accepted closed-form "
    "power formula for the Wilcoxon signed-rank test is unavailable. "
    "The actual hypothesis test for RQ1 will remain the one-sided "
    "Wilcoxon signed-rank test."
)
```

## **1.5 Sample Size Determination for RQ2**

### **1.5.1 Statistical Basis**

Research Question 2 investigates which retrieval model provides the best retrieval performance for legal question answering over Indian Supreme Court judgments.

The final experimental design compares the following six retrieval methodologies:

1. TF-IDF
2. BM25
3. E5
4. BGE
5. InLegalBERT
6. Hybrid Retrieval

Each retrieval methodology will be evaluated using the same benchmark set of legal research questions. Therefore, every benchmark question produces retrieval-performance scores for all six methodologies. The observations are consequently related rather than independent, resulting in a **repeated-measures experimental design**.

The primary statistical test selected for RQ2 is the **Friedman test**. The Friedman test is a non-parametric alternative to the one-way repeated-measures analysis of variance and is appropriate for comparing three or more related groups when the assumptions required for parametric analysis cannot be guaranteed.

For each benchmark legal research question, the six retrieval methodologies will be ranked according to the selected retrieval-performance metric. The Friedman test will determine whether statistically significant differences exist among the methodologies.

If the Friedman test produces a statistically significant result, **pairwise Wilcoxon signed-rank post-hoc comparisons** will be conducted to determine which retrieval methodologies differ significantly from one another. The resulting *p*-values will be adjusted using the **Holm correction** to control the family-wise error rate arising from multiple comparisons.

The hypotheses for RQ2 are defined as follows:

- **Null Hypothesis (H₀₂):** There is no statistically significant difference in retrieval performance among TF-IDF, BM25, E5, BGE, InLegalBERT, and Hybrid Retrieval.

- **Alternative Hypothesis (H₁₂):** At least one retrieval methodology exhibits significantly different retrieval performance from the remaining methodologies.

No universally accepted closed-form sample-size formula exists for the Friedman test under a repeated-measures design involving multiple related groups. Therefore, the required sample size for RQ2 is estimated using an asymptotic non-central chi-square approximation based on **Kendall’s coefficient of concordance ($W$)**.

Kendall’s $W$ represents the magnitude of agreement or systematic ranking differences among the retrieval methodologies and serves as the effect-size measure associated with the Friedman test.

In the absence of pilot-derived estimates, the prospective power analysis assumes:

- **Number of retrieval methodologies:** 6
- **Significance level ($\alpha$):** 0.05
- **Target statistical power ($1-\beta$):** 0.80
- **Assumed Kendall’s coefficient of concordance:** $W = 0.20$
- **Prospective sample-size range:** 5–100 benchmark questions

The required sample size is identified as the smallest number of benchmark legal research questions for which the estimated statistical power reaches or exceeds 0.80. The result will subsequently be compared with the sample-size requirements obtained for RQ1, RQ3, and RQ4.

**Table 2. Statistical Design and Power Analysis Parameters for Retrieval Evaluation (RQ2)**

| **Parameter** | **Value / Description** |
|:--------------|:------------------------|
| **Applicable Research Question** | RQ2 – Which retrieval model provides the best retrieval performance for legal question answering over Indian Supreme Court judgments? |
| **Retrieval Methodologies Compared** | TF-IDF, BM25, E5, BGE, InLegalBERT, and Hybrid Retrieval |

| **Target Statistical Power ($1-\beta$)** | 0.80 |
| **Effect-Size Measure** | Kendall's coefficient of concordance ($W$) |
| **Assumed Effect Size** | $W = 0.20$ (conservative prospective assumption) |

The parameters presented in Table 2 define the repeated-measures statistical framework adopted for RQ2. The Friedman test was selected because the retrieval performance of six related retrieval methodologies is evaluated using the same benchmark legal research questions. Sample-size estimation is based on an analytical approximation employing Kendall's coefficient of concordance ($W$) as the effect-size measure. Following identification of the minimum sample size required to achieve the target statistical power, the planned benchmark of 50 legal research questions is assessed for adequacy before conducting the retrieval evaluation.

```python
# ============================================================
# RQ2: ANALYTICAL POWER APPROXIMATION FOR THE FRIEDMAN TEST
# ============================================================


from scipy.stats import chi2, ncx2



# ------------------------------------------------------------
# Step 1: Define the power-analysis parameters
# ------------------------------------------------------------

ALPHA = 0.05
TARGET_POWER = 0.80

# Six retrieval methodologies:
# TF-IDF, BM25, E5, BGE, InLegalBERT, and Hybrid Retrieval
N_METHODOLOGIES = 6

# Conservative prospective effect-size assumption
ASSUMED_KENDALL_W = 0.20

# Candidate numbers of benchmark legal research questions
SAMPLE_SIZES = range(5, 101)

# Planned benchmark sample
PLANNED_SAMPLE_SIZE = 50


# ------------------------------------------------------------
# Step 2: Calculate degrees of freedom and critical value
# ------------------------------------------------------------

DEGREES_OF_FREEDOM = N_METHODOLOGIES - 1

critical_chi_square = chi2.ppf(
    1 - ALPHA,
    df=DEGREES_OF_FREEDOM
)


# ------------------------------------------------------------
# Step 3: Define the analytical power function
# ------------------------------------------------------------

def calculate_friedman_power(
    n_questions,
    n_methodologies,
    kendall_w,
    alpha
):
    """
    Estimate statistical power for the Friedman test using an
    asymptotic non-central chi-square approximation.

    Parameters
    ----------
    n_questions : int
        Number of benchmark legal research questions.

    n_methodologies : int
        Number of related retrieval methodologies.

    kendall_w : float
        Assumed Kendall's coefficient of concordance.

    alpha : float
        Significance level.

    Returns
    -------
    tuple
        Non-centrality parameter and estimated statistical power.
    """

    degrees_of_freedom = n_methodologies - 1

    critical_value = chi2.ppf(
        1 - alpha,
        df=degrees_of_freedom
    )

    # Approximate non-centrality parameter:
    #
    # lambda = N × (k - 1) × W
    #
    # where:
    # N = number of benchmark questions
    # k = number of retrieval methodologies
    # W = assumed Kendall's coefficient of concordance

    noncentrality_parameter = (
        n_questions
        * (n_methodologies - 1)
        * kendall_w
    )

    estimated_power = 1 - ncx2.cdf(
        critical_value,
        df=degrees_of_freedom,
        nc=noncentrality_parameter
    )

    return noncentrality_parameter, estimated_power


# ------------------------------------------------------------
# Step 4: Calculate power across candidate sample sizes
# ------------------------------------------------------------

power_results = []

for n_questions in SAMPLE_SIZES:

    noncentrality, estimated_power = calculate_friedman_power(
        n_questions=n_questions,
        n_methodologies=N_METHODOLOGIES,
        kendall_w=ASSUMED_KENDALL_W,
        alpha=ALPHA
    )

    power_results.append({
        "Evaluation Questions": n_questions,
        "Non-Centrality Parameter": noncentrality,
        "Estimated Power": estimated_power,
        "Target Power Achieved": estimated_power >= TARGET_POWER
    })


rq2_power_results = pd.DataFrame(power_results)


# ------------------------------------------------------------
# Step 5: Identify the minimum required sample size
# ------------------------------------------------------------

adequate_sample_sizes = rq2_power_results[
    rq2_power_results["Estimated Power"] >= TARGET_POWER
]

if adequate_sample_sizes.empty:
    minimum_required_n_rq2 = None
else:
    minimum_required_n_rq2 = int(
        adequate_sample_sizes.iloc[0]["Evaluation Questions"]
    )


# ------------------------------------------------------------
# Step 6: Extract power for the planned sample of 50 questions
# ------------------------------------------------------------

planned_sample_power_rq2 = float(
    rq2_power_results.loc[
        rq2_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE,
        "Estimated Power"
    ].iloc[0]
)


# ------------------------------------------------------------
# Step 7: Validation checks
# ------------------------------------------------------------

assert 0 < ALPHA < 1, (
    "Alpha must be between 0 and 1."
)

assert 0 < TARGET_POWER < 1, (
    "Target power must be between 0 and 1."
)

assert 0 < ASSUMED_KENDALL_W <= 1, (
    "Kendall's W must be greater than 0 and no more than 1."
)

assert N_METHODOLOGIES >= 3, (
    "The Friedman test requires at least three related groups."
)

assert PLANNED_SAMPLE_SIZE in rq2_power_results[
    "Evaluation Questions"
].values, (
    "The planned sample size is not included in the candidate range."
)

assert rq2_power_results["Estimated Power"].between(0, 1).all(), (
    "All estimated power values must be between 0 and 1."
)

assert rq2_power_results["Estimated Power"].is_monotonic_increasing, (
    "Estimated power should increase as sample size increases."
)

assert rq2_power_results.isna().sum().sum() == 0, (
    "The results table contains missing values."
)


# ------------------------------------------------------------
# Step 8: Display the most relevant rows
# ------------------------------------------------------------

if minimum_required_n_rq2 is not None:

    lower_limit = max(
        min(SAMPLE_SIZES),
        minimum_required_n_rq2 - 3
    )

    upper_limit = minimum_required_n_rq2 + 3

    nearby_rows = rq2_power_results[
        rq2_power_results["Evaluation Questions"].between(
            lower_limit,
            upper_limit
        )
    ]

    planned_row = rq2_power_results[
        rq2_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE
    ]

    display_table = pd.concat(
        [nearby_rows, planned_row],
        ignore_index=True
    ).drop_duplicates(
        subset=["Evaluation Questions"]
    )

else:
    display_table = rq2_power_results.tail(10)


display(
    display_table.style
    .hide(axis="index")
    .format({
        "Non-Centrality Parameter": "{:.3f}",
        "Estimated Power": "{:.3f}"
    })
    .set_caption(
        "RQ2 Analytical Power Approximation for the Friedman Test"
    )
    .set_properties(**{
        "text-align": "center",
        "vertical-align": "middle"
    })
)


# ------------------------------------------------------------
# Step 9: Print the final sample-size conclusion
# ------------------------------------------------------------

print("=" * 72)
print("RQ2 SAMPLE-SIZE DETERMINATION")
print("=" * 72)

print(
    f"Number of retrieval methodologies : "
    f"{N_METHODOLOGIES}"
)

print(
    f"Degrees of freedom                 : "
    f"{DEGREES_OF_FREEDOM}"
)

print(
    f"Primary statistical test           : "
    f"Friedman test"
)

print(
    f"Post-hoc analysis                  : "
    f"Wilcoxon signed-rank tests with Holm correction"
)

print(
    f"Significance level                 : "
    f"{ALPHA:.2f}"
)

print(
    f"Target statistical power           : "
    f"{TARGET_POWER:.2f}"
)

print(
    f"Assumed Kendall's W                : "
    f"{ASSUMED_KENDALL_W:.2f}"
)

print(
    f"Critical chi-square value          : "
    f"{critical_chi_square:.3f}"
)


if minimum_required_n_rq2 is not None:

    print(
        f"\nMinimum sample size achieving power ≥ "
        f"{TARGET_POWER:.2f}: "
        f"{minimum_required_n_rq2} questions"
    )

    print(
        f"Estimated power with "
        f"{PLANNED_SAMPLE_SIZE} questions       : "
        f"{planned_sample_power_rq2:.3f}"
    )

    if PLANNED_SAMPLE_SIZE >= minimum_required_n_rq2:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions exceeds "
            f"the estimated minimum sample-size requirement for RQ2."
        )

    else:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions does not meet "
            f"the estimated minimum sample-size requirement for RQ2."
        )

else:

    print(
        "\nThe target statistical power was not achieved "
        "within the candidate sample-size range."
    )


print(
    "\nNote: This result is based on an asymptotic non-central "
    "chi-square approximation and an assumed Kendall's W of 0.20. "
    "It represents a prospective planning estimate rather than "
    "an exact sample-size calculation."
)
```

### **Observations**

The prospective power analysis for **RQ2** was performed using an analytical approximation for the Friedman test based on **Kendall's coefficient of concordance ($W$)**. The analysis assumed a significance level of **α = 0.05**, a target statistical power of **0.80**, and a conservative effect-size assumption of **$W = 0.20$**.

The estimated statistical power increased monotonically with the number of benchmark legal research questions. The minimum sample size required to achieve the target statistical power was **13 evaluation questions**, at which the estimated power reached **0.806**.

For the planned benchmark dataset of **50 legal research questions**, the estimated statistical power was **1.000**, substantially exceeding the target statistical power of **0.80**.
```python
# ============================================================
# RQ3: ONE-SAMPLE POWER ANALYSIS FOR ANSWER-QUALITY EVALUATION
# ============================================================

import pandas as pd
from statsmodels.stats.power import TTestPower
from IPython.display import display


# ------------------------------------------------------------
# Step 1: Define the power-analysis parameters
# ------------------------------------------------------------

ALPHA = 0.05
TARGET_POWER = 0.80

# Conservative prospective standardized one-sample effect size
ASSUMED_EFFECT_SIZE = 0.50

# One-sided hypothesis:
# NYAYaRAG answer quality is greater than the benchmark
ALTERNATIVE = "larger"

# Candidate numbers of benchmark legal research questions
SAMPLE_SIZES = range(5, 101)

# Planned benchmark sample
PLANNED_SAMPLE_SIZE = 50


# ------------------------------------------------------------
# Step 2: Initialize the one-sample power-analysis model
# ------------------------------------------------------------

power_analysis = TTestPower()


# ------------------------------------------------------------
# Step 3: Calculate estimated power across candidate sample sizes
# ------------------------------------------------------------

power_results = []

for n_questions in SAMPLE_SIZES:

    estimated_power = power_analysis.power(
        effect_size=ASSUMED_EFFECT_SIZE,
        nobs=n_questions,
        alpha=ALPHA,
        alternative=ALTERNATIVE
    )

    power_results.append({
        "Evaluation Questions": n_questions,
        "Assumed Effect Size": ASSUMED_EFFECT_SIZE,
        "Estimated Power": estimated_power,
        "Target Power Achieved": estimated_power >= TARGET_POWER
    })


rq3_power_results = pd.DataFrame(power_results)


# ------------------------------------------------------------
# Step 4: Identify the minimum required sample size
# ------------------------------------------------------------

adequate_sample_sizes = rq3_power_results[
    rq3_power_results["Estimated Power"] >= TARGET_POWER
]

if adequate_sample_sizes.empty:
    minimum_required_n_rq3 = None
else:
    minimum_required_n_rq3 = int(
        adequate_sample_sizes.iloc[0]["Evaluation Questions"]
    )


# ------------------------------------------------------------
# Step 5: Confirm the minimum sample size analytically
# ------------------------------------------------------------

analytical_required_n_rq3 = power_analysis.solve_power(
    effect_size=ASSUMED_EFFECT_SIZE,
    power=TARGET_POWER,
    alpha=ALPHA,
    alternative=ALTERNATIVE
)


# ------------------------------------------------------------
# Step 6: Extract power for the planned sample of 50 questions
# ------------------------------------------------------------

planned_sample_power_rq3 = float(
    rq3_power_results.loc[
        rq3_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE,
        "Estimated Power"
    ].iloc[0]
)


# ------------------------------------------------------------
# Step 7: Validation checks
# ------------------------------------------------------------

assert 0 < ALPHA < 1, (
    "Alpha must be between 0 and 1."
)

assert 0 < TARGET_POWER < 1, (
    "Target power must be between 0 and 1."
)

assert ASSUMED_EFFECT_SIZE > 0, (
    "The assumed standardized effect size must be greater than 0."
)

assert PLANNED_SAMPLE_SIZE in rq3_power_results[
    "Evaluation Questions"
].values, (
    "The planned sample size is not included in the candidate range."
)

assert rq3_power_results["Estimated Power"].between(0, 1).all(), (
    "All estimated power values must be between 0 and 1."
)

assert rq3_power_results["Estimated Power"].is_monotonic_increasing, (
    "Estimated power should increase as sample size increases."
)

assert rq3_power_results.isna().sum().sum() == 0, (
    "The results table contains missing values."
)

if minimum_required_n_rq3 is not None:
    assert minimum_required_n_rq3 == int(
        analytical_required_n_rq3
        if analytical_required_n_rq3.is_integer()
        else analytical_required_n_rq3 + 1
    ), (
        "The iterative and analytical sample-size estimates do not match."
    )


# ------------------------------------------------------------
# Step 8: Display the most relevant rows
# ------------------------------------------------------------

if minimum_required_n_rq3 is not None:

    lower_limit = max(
        min(SAMPLE_SIZES),
        minimum_required_n_rq3 - 3
    )

    upper_limit = minimum_required_n_rq3 + 3

    nearby_rows = rq3_power_results[
        rq3_power_results["Evaluation Questions"].between(
            lower_limit,
            upper_limit
        )
    ]

    planned_row = rq3_power_results[
        rq3_power_results["Evaluation Questions"]
        == PLANNED_SAMPLE_SIZE
    ]

    display_table = pd.concat(
        [nearby_rows, planned_row],
        ignore_index=True
    ).drop_duplicates(
        subset=["Evaluation Questions"]
    )

else:
    display_table = rq3_power_results.tail(10)


display(
    display_table.style
    .hide(axis="index")
    .format({
        "Assumed Effect Size": "{:.2f}",
        "Estimated Power": "{:.3f}"
    })
    .set_caption(
        "RQ3 One-Sample Power Analysis for NYAYaRAG Answer-Quality Evaluation"
    )
    .set_properties(**{
        "text-align": "center",
        "vertical-align": "middle"
    })
)


# ------------------------------------------------------------
# Step 9: Print the final sample-size conclusion
# ------------------------------------------------------------

print("=" * 72)
print("RQ3 SAMPLE-SIZE DETERMINATION")
print("=" * 72)

print(
    f"System evaluated                    : "
    f"NYAYaRAG"
)

print(
    f"Primary statistical test           : "
    f"One-sample Wilcoxon signed-rank test"
)

print(
    f"Sample-size approximation           : "
    f"One-sample normal approximation"
)

print(
    f"Direction of hypothesis             : "
    f"One-sided"
)

print(
    f"Significance level                  : "
    f"{ALPHA:.2f}"
)

print(
    f"Target statistical power            : "
    f"{TARGET_POWER:.2f}"
)

print(
    f"Assumed standardized effect size    : "
    f"{ASSUMED_EFFECT_SIZE:.2f}"
)


if minimum_required_n_rq3 is not None:

    print(
        f"\nAnalytical required sample size      : "
        f"{analytical_required_n_rq3:.3f}"
    )

    print(
        f"Minimum whole-number sample size     : "
        f"{minimum_required_n_rq3} questions"
    )

    print(
        f"Estimated power with "
        f"{PLANNED_SAMPLE_SIZE} questions        : "
        f"{planned_sample_power_rq3:.3f}"
    )

    if PLANNED_SAMPLE_SIZE >= minimum_required_n_rq3:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions exceeds "
            f"the estimated minimum sample-size requirement for RQ3."
        )

    else:

        print(
            f"\nConclusion: The planned sample of "
            f"{PLANNED_SAMPLE_SIZE} benchmark questions does not meet "
            f"the estimated minimum sample-size requirement for RQ3."
        )

else:

    print(
        "\nThe target statistical power was not achieved "
        "within the candidate sample-size range."
    )


print(
    "\nNote: This calculation uses a one-sample normal approximation "
    "for prospective planning. The final hypothesis test for RQ3 "
    "will remain the one-sample Wilcoxon signed-rank test."
)
```

### **Observations**

The prospective power analysis for **RQ3** was conducted using a one-sample normal approximation for the planned one-sample Wilcoxon signed-rank test. The analysis assumed a significance level of **$\alpha = 0.05$**, a target statistical power of **$1-\beta = 0.80$**, a one-sided alternative hypothesis, and a conservative standardized effect size of **$d = 0.50$**.

The estimated statistical power increased progressively as the number of benchmark legal research questions increased. The analytical calculation produced a required sample size of **26.138**, which was rounded upward to the next whole number. Accordingly, the minimum sample size required to achieve the target power was **27 benchmark questions**, at which the estimated statistical power was **0.812**.

For the planned evaluation dataset of **50 benchmark legal research questions**, the estimated statistical power was **0.967**, which substantially exceeds the required power of **0.80**. This indicates that the planned sample is sufficiently large to evaluate whether the median NYAYaRAG answer-quality score is significantly greater than the predefined benchmark.

Accordingly, the planned sample of **50 benchmark questions** exceeds the minimum sample-size requirement for RQ3 and is considered adequate for the subsequent one-sample Wilcoxon signed-rank analysis.

## **1.7 Sample Size Determination for NYAYaRAG Abstention Evaluation (RQ4)**

### **1.7.1 Statistical Basis**

Research Question 4 evaluates whether **NYAYaRAG appropriately abstains from generating an answer when a legal research question cannot be answered from the retrieved Indian Supreme Court judgments**. In such cases, the system should acknowledge that sufficient supporting information is unavailable rather than generate an unsupported or potentially hallucinated answer.

Each unanswerable benchmark question produces a binary outcome:

- **Correct abstention:** NYAYaRAG appropriately declines to answer because the retrieved judgments do not contain sufficient supporting information.
- **Incorrect abstention decision:** NYAYaRAG generates an unsupported answer despite the absence of adequate evidence in the retrieved judgments.

The outcome variable is therefore the **proportion of correct abstention decisions**. As each question produces either a correct or incorrect abstention outcome, RQ4 follows a **one-sample binomial evaluation design**.

The primary inferential procedure selected for RQ4 is the **exact one-sided binomial test**. This test will determine whether the observed correct-abstention proportion is significantly greater than a predefined benchmark proportion.

Because no universally established correct-abstention benchmark currently exists for legal retrieval-augmented generation systems, three prospective benchmark proportions were examined:

- **50% benchmark ($p_0 = 0.50$):** Determines whether NYAYaRAG performs better than a basic majority-level threshold.
- **60% benchmark ($p_0 = 0.60$):** Represents a moderate minimum acceptable level of correct abstention performance.
- **70% benchmark ($p_0 = 0.70$):** Represents a comparatively demanding minimum acceptable level of correct abstention performance.

For each benchmark, the hypotheses are expressed generally as:

- **Null Hypothesis ($H_{04}$):** The true proportion of correct abstention decisions is less than or equal to the specified benchmark proportion.

- **Alternative Hypothesis ($H_{14}$):** The true proportion of correct abstention decisions is greater than the specified benchmark proportion.

Mathematically:

$$
H_{04}: p \leq p_0
$$

$$
H_{14}: p > p_0
$$

where:

- $p$ represents the true correct-abstention proportion of NYAYaRAG; and
- $p_0$ represents the selected benchmark proportion of 0.50, 0.60, or 0.70.

The prospective power analysis assumes an expected correct-abstention proportion of **$p_1 = 0.85$**, representing the anticipated performance of NYAYaRAG under the alternative hypothesis. The remaining statistical assumptions are:

- **Significance level ($\alpha$):** 0.05
- **Target statistical power ($1-\beta$):** 0.80
- **Expected correct-abstention proportion ($p_1$):** 0.85
- **Alternative hypothesis:** One-sided
- **Candidate sample-size range:** 5–100 unanswerable benchmark questions

For each benchmark proportion, the minimum required sample size was defined as the smallest number of unanswerable benchmark questions for which the exact statistical power reached or exceeded **0.80**.

### **Table 4. Comparison of Statistical Design and Sample-Size Requirements for Alternative Abstention Benchmarks (RQ4)**

| **Parameter** | **50% Benchmark** | **60% Benchmark** | **70% Benchmark** |
|:---|:---:|:---:|:---:|
| **Benchmark proportion ($p_0$)** | 0.50 | 0.60 | 0.70 |
| **Expected proportion ($p_1$)** | 0.85 | 0.85 | 0.85 |
| **Significance level ($\alpha$)** | 0.05 | 0.05 | 0.05 |
| **Target power ($1-\beta$)** | 0.80 | 0.80 | 0.80 |
| **Primary statistical test** | Exact one-sided binomial test | Exact one-sided binomial test | Exact one-sided binomial test |
| **Minimum required sample size** | **13** | **21** | **49** |
| **Critical correct abstentions** | **10 of 13** | **17 of 21** | **40 of 49** |
| **Estimated power at minimum sample size** | 0.882 | 0.803 | 0.809 |
| **Interpretation** | Tests whether performance exceeds a basic 50% threshold | Tests whether performance exceeds a moderate 60% threshold | Tests whether performance exceeds a demanding 70% threshold |

The prospective analysis demonstrates that the minimum required sample size increases as the benchmark becomes more demanding. A benchmark of **50%** requires 13 unanswerable questions, whereas benchmarks of **60%** and **70%** require 21 and 49 unanswerable questions, respectively.

The **50% benchmark** provides the smallest sample-size requirement but represents a comparatively weak standard for a legal question-answering system. The **70% benchmark** provides a stronger safety-oriented standard but substantially increases the required evaluation sample. The **60% benchmark** offers an intermediate and practically feasible threshold while still requiring NYAYaRAG to demonstrate correct-abstention performance meaningfully above a basic majority-level benchmark.

Subject to the final methodological justification, the **60% benchmark** may therefore be adopted for RQ4. Under this design, the final hypotheses would be:

$$
H_{04}: p \leq 0.60
$$

$$
H_{14}: p > 0.60
$$

The corresponding minimum required sample size would be **21 unanswerable benchmark legal research questions**. A planned sample of **25–30 unanswerable questions** would exceed this minimum requirement and provide a modest allowance for exclusions, ambiguous classifications, or unusable evaluation instances.

```python
# ============================================================
# COMPARISON OF RQ4 SAMPLE SIZE FOR DIFFERENT BENCHMARKS
# ============================================================

import pandas as pd
from scipy.stats import binom

# Common parameters
EXPECTED_PROPORTION = 0.85
ALPHA = 0.05
TARGET_POWER = 0.80

# Compare different benchmark proportions
NULL_PROPORTIONS = [0.50, 0.60, 0.70]

SAMPLE_SIZES = range(5, 101)


def calculate_exact_power(n, p0, p1, alpha):

    critical_successes = None

    for k in range(n + 1):

        if binom.sf(k - 1, n, p0) <= alpha:
            critical_successes = k
            break

    if critical_successes is None:
        return None, 0.0

    power = binom.sf(
        critical_successes - 1,
        n,
        p1
    )

    return critical_successes, power


summary = []

for p0 in NULL_PROPORTIONS:

    minimum_n = None
    critical = None
    achieved_power = None

    for n in SAMPLE_SIZES:

        critical_successes, power = calculate_exact_power(
            n,
            p0,
            EXPECTED_PROPORTION,
            ALPHA
        )

        if power >= TARGET_POWER:

            minimum_n = n
            critical = critical_successes
            achieved_power = power
            break

    summary.append({
        "Benchmark Proportion (p0)": p0,
        "Expected Proportion (p1)": EXPECTED_PROPORTION,
        "Minimum Sample Size": minimum_n,
        "Critical Correct Abstentions": critical,
        "Estimated Power": achieved_power
    })

summary_df = pd.DataFrame(summary)

display(
    summary_df.style
    .hide(axis="index")
    .format({
        "Benchmark Proportion (p0)": "{:.2f}",
        "Expected Proportion (p1)": "{:.2f}",
        "Estimated Power": "{:.3f}"
    })
    .set_caption("Comparison of Sample Size Requirements for Different Abstention Benchmarks")
)
```

### **Observations**

The exact one-sided binomial power analysis compared three possible benchmark proportions for evaluating the abstention performance of NYAYaRAG. In all three scenarios, the expected correct-abstention proportion was fixed at **$p_1 = 0.85$**, the significance level was set at **$\alpha = 0.05$**, and the target statistical power was **$1-\beta = 0.80$**.

For the **50% benchmark** ($p_0 = 0.50$), the minimum required sample size was **13 unanswerable questions**. At least **10 correct abstentions out of 13 questions** would be required to reject the null hypothesis. The estimated statistical power at this sample size was **0.882**. Although this design requires the smallest evaluation sample, a 50% benchmark represents a relatively weak standard because it only tests whether NYAYaRAG performs better than a basic majority-level threshold.

For the **60% benchmark** ($p_0 = 0.60$), the minimum required sample size was **21 unanswerable questions**. At least **17 correct abstentions out of 21 questions** would be required to reject the null hypothesis. The estimated statistical power was **0.803**, which marginally exceeds the predefined target of 0.80. This benchmark represents a moderate minimum performance threshold and provides a balance between statistical adequacy and practical feasibility.

For the **70% benchmark** ($p_0 = 0.70$), the minimum required sample size was **49 unanswerable questions**. At least **40 correct abstentions out of 49 questions** would be required to reject the null hypothesis. The estimated statistical power was **0.809**. This benchmark imposes a more demanding abstention-performance standard but substantially increases the number of unanswerable questions required for evaluation.

The results demonstrate that the required sample size increases as the benchmark proportion approaches the expected performance of 85%. The difference between the null benchmark and the expected performance is largest for the 50% benchmark and smallest for the 70% benchmark. Consequently, fewer questions are required to detect improvement over 50%, whereas a much larger sample is required to establish that the true correct-abstention rate exceeds 70%.

Considering both methodological rigour and practical feasibility, the **60% benchmark** provides an appropriate intermediate threshold for the present study. Under this design, a minimum of **21 unanswerable benchmark legal research questions** is required. A planned sample of approximately **25–30 unanswerable questions** would exceed this minimum requirement and provide an allowance for exclusions, ambiguous classifications, or unusable evaluation instances.
